```python
import json
from pathlib import Path

with open("../data/raw/products_raw.json", "r", encoding="utf-8") as f:
    products = json.load(f)

print(f"Loaded {len(products)} products.")
```

```python
sample = products[0]
print("Top-level keys:", sorted(sample.keys()))
print()
print("sugars_100g value:", sample.get("nutriments", {}).get("sugars_100g"))
```

```python
for i in range(8):
    p = products[i]
    print(f"Product {i}: {p.get('product_name', 'NO NAME')}")
    print(f"  Keys: {sorted(p.keys())}")
    print()
```

```python
missing_key = 0
empty_string = 0
has_text = 0

for p in products:
    if "ingredients_text" not in p:
        missing_key += 1
    elif p.get("ingredients_text") == "":
        empty_string += 1
    else:
        has_text += 1

print(f"Missing key entirely: {missing_key}")
print(f"Empty string: {empty_string}")
print(f"Has actual text: {has_text}")
```

```python
sugar_values = []

for p in products:
    sugar = p.get("nutriments", {}).get("sugars_100g")
    if sugar is not None:
        sugar_values.append(sugar)

min_sugar = sugar_values[0]
max_sugar = sugar_values[0]
total = 0

for val in sugar_values:
    if val < min_sugar:
        min_sugar = val
    if val > max_sugar:
        max_sugar = val
    total += val

mean_sugar = total / len(sugar_values)

print(f"Count with sugar data: {len(sugar_values)} out of {len(products)}")
print(f"Min sugar per 100g: {min_sugar}")
print(f"Max sugar per 100g: {max_sugar}")
print(f"Mean sugar per 100g: {mean_sugar:.2f}")
```

```python
fields_to_check = ["sugars_100g", "fat_100g", "fiber_100g", "salt_100g", "proteins_100g", "energy-kcal_100g"]

for field in fields_to_check:
    present_count = 0
    for p in products:
        value = p.get("nutriments", {}).get(field)
        if value is not None:
            present_count += 1
    pct = (present_count / len(products)) * 100
    print(f"{field}: {present_count}/{len(products)} present ({pct:.1f}%)")
```

```python
grade_counts = {}

for p in products:
    grade = p.get("nutrition_grades")
    grade_counts[grade] = grade_counts.get(grade, 0) + 1

for grade, count in sorted(grade_counts.items(), key=lambda x: str(x[0])):
    print(f"Grade {grade}: {count} products")
```
